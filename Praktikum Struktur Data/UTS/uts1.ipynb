```python
class siswa:
    def __init__(self, nomor_induk, nilai):
        self.nomor_induk = nomor_induk
        self.nilai = nilai
        
def ujian(siswa_list):
    siswa_terpanggil = []
    nilai_sempurna = []
    
    for siswa in siswa_list:
        siswa_terpanggil.append(siswa)
        print("Semua dengan nomor induk", siswa.nomor_induk, "sedang dalam wawancara")
        nilai = int(input("Masukan nilai ujian: "))
        siswa.nilai = nilai
        
        if nilai == 100:
            nilai_sempurna.append(siswa)
    
    print("Hasil ujian")
    for siswa in siswa_terpanggil:
        if siswa.nilai <= 60:
            print("Siswa dengan nomor induk", siswa.nomor_induk, "tidak lulus")
        else:
            print("Siswa dengan nomor induk", siswa.nomor_induk, "lulus")
                
    if nilai_sempurna:
        print("Siswa nilai sempurna")
        for siswa in nilai_sempurna:
            print("Nomor induk", siswa.nomor_induk)

def ururtan(siswa_list):
    return sorted(siswa_list, key=lambda x: x.nomor_induk)

if __name__ == "__main__":
    jumlah_siswa = int(input("Masukan jumlah siswa: "))
    siswa_list = []
    
    for i in range(jumlah_siswa):
        nomor_induk = int(input("Masukan nomor induk siswa ke-{}: ".format(i+1)))
        siswa_list.append(siswa(nomor_induk, 0))
        
    siswa_list = ururtan(siswa_list)
    ujian(siswa_list)
```
