Tree sebelum proses hapus: 8 3 15 12 7

Tree setelah proses hapus: 8 3 15 7

Tree setelah proses insert: 8 3 5 15 7 6

```python
class Node:
    def __init__(self, data): 
        self.left = None
        self.right = None
        self.data = data
        
#Insert Node
    def insert(self, data):
        if self.data:
            if data < self.data:
                if  self.left is None:
                    self.left = Node(data)
                else:
                    self.left.insert(data)
            elif data> self.data: 
                if  self.right is None:
                    self.right = Node(data)
                else:
                    self.right.insert(data)
        else:
            self.data = data
                        
def inorder(temp):
    if temp is None:
        return
    inorder(temp.left)
    print(temp.data, end=" ")
    inorder(temp.right)
    
def deleteDeepest(root, d_node):
    q = []
    q.append(root)
    while len(q):
        temp = q.pop(0)
        if temp is d_node:
            temp = None
            return
        if temp.right:
            if temp.right is d_node:
                temp.right = None
                return
            else:
                q.append(temp.right)
        if temp.left:
            if temp.left is d_node:
                temp.left = None
                return
            else:
                q.append(temp.left)
                
def deletion(root, data):
    if root is None:
        return None
    if root.data == data and root.left is None and root.right is None:
        return None

    q = [root]
    temp = None
    key_node = None
    while q:
        temp = q.pop(0)
        if temp.data == data:
            key_node = temp
        if temp.left:
            q.append(temp.left)
        if temp.right:
            q.append(temp.right)

    if key_node:
        x = temp.data
        key_node.data = x
        deleteDeepest(root, temp)
    return root


if __name__ == '__main__':
    root = Node(8)
    root.insert(3)
    root.insert(15)
    root.insert(12)
    root.insert(7)
    
    print("Tree sebelum proses hapus:", end=" ")
    inorder(root)
    
    data = 12
    root = deletion(root, data)
    print("\nTree setelah proses hapus:", end=" ")
    inorder(root)

    data = [5, 6]
    for d in data:
        root.insert(d)
    print("\nTree setelah proses insert:", end=" ")
    inorder(root)

```

2.Memperbaiki Progam agar input sesuai infomasi berikut: ("Paku", 5) ("Palu", 10) ("Gergaji", 15) ("Skop", 20) ("Pasah", 25) sampai menghasilkan output sebagai berikut: 20 25 10

```python
class Node:
    def __init__(self, key, value):
        self.key = key
        self.value = value
        self.next = None

class separatechaining:
    def __init__(self, size):
        self.size = size
        self.table = [None] * size;

    def hash_function(self, key):
        return hash(key) % self.size;

    def insert(self, key, value):
        index = self.hash_function(key)
        if self.table[index] is None:
            self.table[index] = Node(key, value)
        else:
            current = self.table[index]
            while current.next:
                current = current.next
            current.next = Node(key, value)

    def search(self, key):
        index = self.hash_function(key)
        current = self.table[index]
        while current:
            if current.key == key:
                return current.value
            current = current.next
        return None

if __name__ == "__main__":
    hash_table = separatechaining(5)
    hash_table.insert("Paku", 5)
    hash_table.insert("Palu", 10)
    hash_table.insert("Gergaji", 15)
    hash_table.insert("Skop", 20)
    hash_table.insert("Pasah", 25)

    print("Hasil pencarian:", end=" ")
    print(hash_table.search("Skop"), end=" ")
    print(hash_table.search("Pasah"), end=" ")
    print(hash_table.search("Palu"))
```

3.Diketahui informasi berikut :
Ani berhutang ke Budi sebesar 5.000
Ani juga memiliki hutang ke Doni sebesar 10.000
Budi memiliki hutang ke Cahyo sebesar 10.000
Cahyo berhutang ke Ani sebesar 1.000
Cahyo berhutang ke Eko sebesar 10.000
Doni memiliki hutang ke Cahyo sebesar 5.000
Eko memiliki hutang kepada Budi sebesar 5.000 dan kepada Doni sebesar 500
Dari data diatas transasi minimum yang dapat dilakukan untuk menyelesaikan persoalan keuangan tersebut.

```python
def transaksi_minimum(hutang):
    # Inisialisasi daftar transaksi
    transaksi = []

    # Selama masih ada hutang yang harus dibayarkan
    while hutang:
        # Mengurutkan hutang berdasarkan jumlah hutang
        hutang.sort(key=lambda x: x[1])

        # Mengambil hutang dengan jumlah paling kecil
        peminjam, jumlah = hutang.pop(0)

        # Cek apakah ada pemberi hutang yang bisa membayarkan hutang
        for i in range(len(hutang)):
            pemberi, sisa = hutang[i]
            if sisa >= jumlah:
                # Jika pemberi hutang cukup membayar, tambahkan transaksi
                transaksi.append((pemberi, peminjam, jumlah))
                hutang[i] = (pemberi, sisa - jumlah)
                break
            else:
                # Jika pemberi hutang tidak cukup, tambahkan transaksi sebagian
                transaksi.append((pemberi, peminjam, sisa))
                jumlah -= sisa
                hutang[i] = (pemberi, 0)

    return transaksi

if __name__ == "__main__":
    # Informasi hutang
    hutang = [
        ("Ani", 5000),
        ("Ani", 10000),
        ("Budi", 10000),
        ("Cahyo", 1000),
        ("Cahyo", 10000),
        ("Doni", 5000),
        ("Eko", 5000),
        ("Eko", 500)
    ]

    # Melakukan transaksi minimum
    hasil_transaksi = transaksi_minimum(hutang)

    # Menampilkan transaksi minimum
    print("Transaksi Minimum:")
    for transaksi in hasil_transaksi:
        print(f"{transaksi[0]} membayar kepada {transaksi[1]} sebesar {transaksi[2]}")
```
